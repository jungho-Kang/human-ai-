```python
data = [[1.40, np.nan],
        [7.10, -4.5],
        [np.nan, np.nan],
        [0.75, -1.3]]
```

```python
data
```

```python
df = pd.DataFrame(data, columns = ['one','two'], index = ['a','b','c','d'])
df
```

```python
# 행 방향으로 열의 합을 계산 (nan은 배제하고 계산됨)
df.sum(axis=0)
```

```python
# 열 방향으로 각 행의 합을 계산
df.sum(axis=1)
```

```python
# 특정 열 또는 행의 값 계산
df['one'].sum()
```

```python
df.loc['b'].sum()
```

```python
df.mean(axis=0)['two']
```

```python
df.mean(axis=1, skipna=False)
```

```python
df
```

```python
one_mean = df.mean(axis=0)['one']
two_min = df.min(axis=0)['two']
```

```python
df['one'] = df['one'].fillna(value=one_mean)
df['two'] = df['two'].fillna(value=two_min)
df
```

```python
# 상관계수 구하기
df2 = pd.DataFrame(np.random.randn(6,4),
                   columns = ['a','b','c','d'],
                   index = pd.date_range('20250101', periods=6))
df2
```

```python
# a열과 b열의 상관계수 산출하기 (피어슨상관계수로 계산)
df2['b'].corr(df2['c'])
```

```python
df2.corr()
```

```python
# 정렬 및 기타 함수
dates = df2.index
random_dates = np.random.permutation(dates)
df2 = df2.reindex(index=random_dates, columns=['a','b','c','d'])
df2
```

```python
# index 기준으로 오름차순 정렬
df2.sort_index(axis=0)
```

```python
# 컬럼 기준으로 오름차순 정렬
df2.sort_index(axis=1)
```

```python
# index 기준으로 내림차순 정렬
df2.sort_index(axis=0, ascending=False)
```

```python
# value 기준으로 오름차순 정렬
df2.sort_values(by='a')
```

```python
# value 기준으로 내림차순 정렬
df2.sort_values(by='b', ascending=False)
```
