## 통계 함수
numpy의 array에 적용되는 통계함수 정리
- sum : 전체 성분의 합을 계산

```python
arr = np.random.randn(5,4)
arr
```

```python
# arr array의 모든 값에 대하여 sum을 수행
arr.sum()
```

```python
arr.mean()
```

```python
# 행 방향으로 열 값들을 전체 sum을 수행 (각 열의 합 계산)
arr.sum(axis=0)
```

```python
# 열 방향으로 행의 값들을 전체 sum을 수행 (각 행의 합 계산)
arr.sum(axis=1)
```

```python
# 조건부 연산
arr > 0
```

```python
# 0보다 큰 값의 개수 출력
(arr > 0).sum()
```

```python
# 0보다 큰 값들 모두 더함
arr[arr > 0].sum()
```

## 정렬함수 및 기타함수

```python
arr = np.random.randn(10)
arr
```

```python
# 오름차순 정렬
np.sort(arr)
```

```python
# 내림차순 정렬
np.sort(arr)[::-1]
```

```python
# 2차원 array 정렬
arr = np.random.randn(5,3)
arr
```

2차원 array를 정의, axis=0을 설정하면 행 방향으로 오름차순 정렬, axis=1을 설정하면 열 방향으로 오름차순 정렬

```python
# 행 방향으로 열의 값을 오름차순 정렬
np.sort(arr, axis=0)
```

```python
# 열 방향으로 행의 값을 오름차순 정렬
np.sort(arr, axis=1)
```

```python
# 길이가 아주 긴 array가 있을 때, 상위 5%에 위치하는 값 찾기
# 1. 데이터 생성 및 내림차순 
large_arr = np.random.randn(150)
large_arr
```

```python
# 내림차순 정렬
np.sort(large_arr)[::-1]
```

```python
# 상위 5%에 해당하는 값 찾기
np.sort(large_arr)[::-1][int(0.05*len(large_arr))]
# 내림차순 된 array에서 0.05*150을 수행하여 int 값으로 변경, 계산된 7번째 자리 값이 5%의 값이다.
```

```python
# 중복된 값 제거하기 - unique()
names = np.array(['aaa','aaa','aaa','fff','fff','ddd','eee'])
ints = np.array([3,3,3,4,4,1,1,2,2,5,5,6,8,8])
```

```python
np.unique(names)
```

```python
np.unique(ints)
```
