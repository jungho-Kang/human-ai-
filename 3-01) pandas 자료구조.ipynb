## 2) DataFrame
DataFrame은 pd.DataFrame() 함수를 사용하여 정의한다. DataFrame

```python
data = {'names': ['Hyunsuk','Hyunsuk','Hyunsuk','Charles','Charles'],
        'year': [2014, 2015, 2016, 2015, 2016],
        'points': [1.5, 1.7, 3.6, 2.4, 2.9]}
df = pd.DataFrame(data)
df
```

```python
df.index
```

```python
df.columns
```

```python
df.values
```

```python
# describe 함수
'''
DataFram df에 대하여 df.describe() 함수를 실행하게 되면,
계산이 가능한 컬럼에 한해서 각 컬럼의 평균, 분산, 최소/최대값 등 기본 통계량을 산출한 결과를 보여준다
이는 데이터셋을 DataFrame 형태로 막 읽어들인 직후에
'''
```

```python
df.describe()
```
